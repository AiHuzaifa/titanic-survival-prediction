```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X)
X_test = scaler.transform(x_test)
```

```python
class Logistic_Regression:
  def __init__(self, lr = 0.01,n_iter = 6000):
    self.lr = lr
    self.n_iter = n_iter
    self.weights = None
    self.base = None

  def sigmoid(self,z):
    return 1/(1+(np.exp(-z)))

  def fit(self,X_train,y_train):
    self.weights = np.zeros(X_train.shape[1])
    self.base = 0
    m = X_train.shape[0]

    for _ in range(1000):
      y_h = np.dot(X_train,self.weights) + self.base
      sig = self.sigmoid(y_h)

      cost = -(y_train*np.log(sig)+(1-(y_train))*np.log(1-sig)).mean()
      if _ % 100 == 0:
        print(cost)

      dw = (X_train.T @ (sig - y_train))/m
      db = (sig-y_train).mean()

      self.weights = self.weights - self.lr*dw
      self.base = self.base - self.lr*db
      if _ % 99 ==0:
        print(self.weights)
        print(self.base)

  def predict(self,x_test):
    y_h = np.dot(x_test,self.weights) + self.base
    return self.sigmoid(y_h)

  def accuracy(self, y_true, y_pred_prob):
    y_pred = np.where(y_pred_prob >= 0.5, 1, 0)
    return np.mean(y_true == y_pred)
```

```python
lg = Logistic_Regression()
lg.fit(X_train,y)
```

```python
y_pred_prob = lg.predict(X_test)
acc = lg.accuracy(Y_test, y_pred_prob)
print(f"Test Accuracy: {acc * 100:.2f}%")
```
